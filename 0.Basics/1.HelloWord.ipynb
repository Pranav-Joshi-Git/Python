When we install python, Python VM also gets installed with it

When we write a code and hit run button
It compiles it to byte code

Byte code is low level & platform independent

Byte code then goes to Python VM and code runs there
When we create a file, write a function in it and try to import it in another
file, then internally a folder gets created, named __pycache__ and it has one
.pyc file in it which python uses internally.


* Code gets compiled into Byte Code
    * Byte code runs faster
    .pyc -> compiled python (frozen binaries)
    __pycache__ 
    hellow_word.cpython-313.pyc
    breakdown -> Source Change & Python Version
    * This works only for imported files
    * And not for top level files

* Python Virtual Machine (PVM)
    * Code loop to iterate byte code
    * PVM is Run time engine of Python
    * Also known as Python Interpreter

* Byte Code is NOT Machine Code
    * Machine code is basically a direct instruction to hardware
    * Byte code is python specific interpretation
    * Variants of Python : cpython(standard implementation), jpython(for java binaries), IronPython, stackless(used in concurrency tasks), PyPy(Performace Oriented)

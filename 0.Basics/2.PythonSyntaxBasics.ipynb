## Python Syntax Basics

A quick reference for core Python syntax and beginner examples.

### First program

**Print output:** `print()` displays the supplied value in the notebook output.

```python
print("Hello World")
```

### Arithmetic operators

**Addition:** `+` adds numeric values.

```python
9 + 2
```

**Mixed operations:** Addition and subtraction have equal precedence and run left to right.

```python
9 - 3 + 1
```

**Parentheses:** The expression inside parentheses runs before multiplication.

```python
(9+2) * 3
```

**Exponentiation:** `**` raises the left value to the power on the right.

```python
8 ** 3
```

**Division:** `/` returns a floating-point result, even when the result is a whole number.

```python
8 / 2
```

**Decimal division:** `/` keeps the fractional part of the result.

```python
9 / 2
```

**Floor division:** `//` divides and rounds the result down.

```python
9 // 2
```

**Modulo:** `%` returns the remainder after division.

```python
9 % 2
```

### Variables

**Assignment:** `=` stores the calculated value under a variable name.

```python
num1 = 9 + 2
num1
```

**Using variables:** Values stored in variables can be combined in later expressions.

```python
n1 = 5
n2 = 6
n3 = n1 + n2
n3
```

### Comments and docstrings

**Single-line comment:** Python ignores text following `#`.

```python
# Single Line Comment
```

**Triple-quoted string:** This creates a multiline string, not a true comment; in specific positions it becomes a docstring.

```python
"""
Multi
Line
Comment
"""
```

### Strings: quotes and escaping

**Mismatched quotes:** This intentionally raises `SyntaxError` because the apostrophe closes the string early.

```python
print('Pranav's Code')
```

**Choose suitable quotes:** Double quotes outside allow the apostrophe to appear normally.

```python
print("Pranav's Code")
```

**Unescaped apostrophe:** This also fails because the apostrophe ends the outer single-quoted string.

```python
print('Pranav's "Code"')
```

**Escape a quote:** `\` escapes the apostrophe; the inner double quotes need no escaping.

```python
print('Pranav\'s "Code"')
```

### String repetition

**Repeat a string:** Multiplying a string by an integer repeats it that many times.

```python
'Pranav ' * 10 
```

### Backslashes and Windows paths

**Unescaped path:** `\U` is treated as a Unicode escape, so this intentionally raises `SyntaxError`.

```python
print('C:\Users\Pranav')
```

**Valid path:** Escape each backslash (`\\`); a raw string like `r'C:\Users\Pranav'` also works.

```python
print('C:\\Users\\Pranav')
print(r'C:\Users\Pranav')
```

### Loops

**Indentation matters:** This intentionally raises `IndentationError` because the loop body is not indented.

```python
for c in "python" :
print(c)
```

### Functions

**Define and call:** `def` creates a reusable function; parentheses call it with the supplied argument.

```python
def display(s):
    print(s)

display("Python")
```

### `repr()`, `str()`, and `print()`

`repr()` returns a developer-focused representation, while `str()` returns a user-friendly string. `print()` displays the `str()` form by default.

```python
# Display and representation

text = "Python"

print(repr(text)) # Includes quotes in the representation
print(str(text))  # User-friendly string
print(text)       # Displays the string
```

### Conditional expressions

A conditional expression chooses one of two values in a single expression. Python writes the successful value first: `value_if_true if condition else value_if_false`.

```python
# Conditional expression (ternary expression)

x = 2
y = 3

message = "yes" if x < y else "no"
print(message)
```

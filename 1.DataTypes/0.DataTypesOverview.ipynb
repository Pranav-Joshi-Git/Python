## Data Types Overview

A short introduction before exploring each Python datatype in depth.

### Object types / data types

Python's commonly used built-in types include:

- **Numbers:** `int`, `float`, `complex`
- **Text and binary data:** `str`, `bytes`
- **Collections:** `list`, `tuple`, `dict`, `set`
- **Logical value:** `bool`
- **No value:** `None`

> Files, decorators, generators, iterators, and metaprogramming are useful Python concepts or objects, but they are not basic datatype categories.

```python
# Number

integer = 10          # int: whole number
decimal = 3.14        # float: decimal number
complex_num = 2 + 3j  # complex: real + imaginary
binary_num = 0b11     # binary representation of 3

print(integer, decimal, complex_num, binary_num)
```

```python
# String

text = "Python"

print(len(text))      # Number of characters
print(text[0])        # First character
print(text[-1])       # Last character
print(text[0:3])      # Slice from index 0 to 2
print(text.upper())   # Call a string method
dir(text)             # Provides list of attributes and methods
```

```python
# Bytes

raw_data = b"ABC"  # bytes store binary data

print(len(raw_data))  # Number of bytes
print(raw_data[0])    # Byte value of 'A' is 65
```

```python
# List

languages = ["Python", "JavaScript"]  # Ordered and mutable

print(languages[0])   # Access by index
print(len(languages)) # Number of items
languages.append("Go")  # Add an item
languages[0] = "Python 3"  # Update an item
print(languages)
```

**List vs tuple:** Lists use `[]` and are mutable (changeable), while tuples use `()` and are immutable (fixed). Use a list when values may change and a tuple when they should remain unchanged.

```python
# Tuple

coordinates = (10, 20)  # Ordered and immutable

print(coordinates[0])    # Access by index
print(len(coordinates))  # Number of items
```

```python
# Dictionary

person = {"name": "Pranav", "age": 25}  # Key-value pairs

print(person["name"])       # Access using a key
print(person.get("city"))   # Returns None when key is missing
person["age"] = 26          # Update a value
```

```python
# Set

unique_numbers = {1, 2, 2, 3}  # Unordered; duplicates are removed

unique_numbers.add(4)           # Add an item
print(unique_numbers)
```

```python
# Boolean

is_learning = True  # Boolean values are True or False

print(is_learning)
print(10 > 5)       # Comparisons produce a boolean
```

```python
# None

result = None         # Represents the absence of a value

print(result)
print(result is None) # Preferred way to check for None
```

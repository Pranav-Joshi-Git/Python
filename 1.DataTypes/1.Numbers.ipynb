## Numbers in Python

Python does not have one single “number” type. It provides several numeric types, and each one solves a different problem:

- `int` for whole numbers and exact integer arithmetic
- `float` for fast approximate decimal arithmetic
- `Decimal` for controlled decimal precision
- `Fraction` for exact rational values
- `complex` for real and imaginary components
- `bool` for logical values; it is also a subclass of `int`

```python
# Common numeric types

integer = 10
decimal_number = 3.14
complex_number = 2 + 3j
boolean_value = True

print(type(integer))
print(type(decimal_number))
print(type(complex_number))
print(type(boolean_value))
```

### Integers: `int`

Integers represent whole numbers. They can be positive, negative, or zero. Python integers have arbitrary precision, so they grow beyond fixed 32-bit or 64-bit limits when required; available memory is the practical limit.

```python
# Integer arithmetic

x = 10
y = 3

print(x + y)   # Addition
print(x - y)   # Subtraction
print(x * y)   # Multiplication
print(x ** y)  # Exponentiation
```

### The same operator can behave differently

Operators call behavior implemented by an object’s type. This is called **operator overloading**. For example, `+` adds numbers but joins strings; `*` multiplies numbers but can repeat a string.

```python
# Operator behavior depends on the object type

print(2 + 3)                 # Numeric addition
print("Python" + " is fun") # String concatenation
print(2 * 3)                 # Numeric multiplication
print("Python " * 3)         # String repetition
```

### Arbitrary-precision integers

Python keeps integer calculations exact even when the result becomes much larger than the range supported by fixed-size integer types in many languages.

```python
# Large integers remain exact

print(100 ** 2)
print(2 ** 100)
print(99999999999999999 + 1)
```

### Division, floor division, and remainder

`/` performs true division and returns a float. `//` rounds the quotient down toward negative infinity. `%` returns the corresponding remainder. `divmod(a, b)` returns both the floor quotient and remainder.

```python
# Division operations

print(5 / 2)          # 2.5: true division
print(5 // 2)         # 2: floor division
print(5 % 2)          # 1: remainder
print(divmod(5, 2))   # (2, 1)
print(-5 // 2)        # -3: rounded down, not toward zero
print(-5 % 2)         # 1: keeps the divmod relationship
```

### Converting numeric values

`int()` and `float()` create values of the requested type. Converting a float with `int()` removes the fractional part by truncating toward zero; it does not round to the nearest integer.

```python
# Numeric conversion

print(int(3.9))     # 3
print(int(-3.9))    # -3
print(float(2))     # 2.0
print(int("42"))   # String to integer
print(float("2.5"))
```

### Binary, octal, and hexadecimal integers

These are different written representations of integer values. The value stored by Python is still an `int`. Prefixes identify the base: `0b` for binary, `0o` for octal, and `0x` for hexadecimal.

```python
# Integer number systems

print(0b11)  # Binary 11 = decimal 3
print(0o10)  # Octal 10 = decimal 8
print(0xFF)  # Hexadecimal FF = decimal 255

print(bin(3))
print(oct(8))
print(hex(255))
```

```python
# Convert strings written in another base

print(int("10000", 2)) # Binary to integer
print(int("32", 8))    # Octal to integer
print(int("FF", 16))   # Hexadecimal to integer
```

### Bitwise operators

Bitwise operators work with the binary bits of integers. They are useful for flags, masks, permissions, and low-level data manipulation.

```python
# Bitwise operators

x = 0b0101  # 5
y = 0b0011  # 3

print(bin(x & y))  # AND: 0b1
print(bin(x | y))  # OR:  0b111
print(bin(x ^ y))  # XOR: 0b110
print(x << 1)      # Left shift: 10
print(x >> 1)      # Right shift: 2
print(~x)          # Bitwise NOT: -6
```

### Floating-point numbers: `float`

Python floats represent real numbers approximately. In CPython they normally use C’s `double` representation, commonly IEEE 754 binary64. A float has fixed precision, unlike an integer. Mixing an `int` with a `float` normally produces a float result.

```python
# Mixed integer and float arithmetic

result = 40 + 2.23

print(result)
print(type(result))
```

### Converting large integers to floats can lose precision

An integer can keep growing exactly, but a float has a limited number of significant binary digits. Once an integer is converted to a float, nearby integer values may become indistinguishable.

```python
# Integer precision versus float precision

large_integer = 2 ** 53 + 1

print(large_integer)        # Exact integer
print(float(large_integer)) # Rounded float representation
print(99999999999999999 * 2.1)
```

### Why decimal arithmetic can look imprecise

Floats are stored in binary. Many simple decimal fractions, including `0.1`, do not have an exact finite binary representation, so Python stores the nearest available float.

```python
# Floating-point representation

print(0.1 + 0.1 + 0.1)
print(0.1 + 0.1 + 0.4)
print(0.1 + 0.1 + 0.1 - 0.3)
```

### Comparing calculated floats

Exact equality may fail after floating-point calculations. `math.isclose()` checks whether two values are sufficiently close within a chosen tolerance.

```python
# Compare floats safely

import math

calculated = 0.1 + 0.1 + 0.1

print(calculated == 0.3)
print(math.isclose(calculated, 0.3))
```

### Rounding floats

`round()` rounds to the requested number of digits. Values exactly halfway between two choices normally use “round half to even,” although binary representation can affect which values are truly halfway.

```python
# Rounding

print(round(3.14159, 2))
print(round(2.5))  # 2: nearest even integer
print(round(3.5))  # 4: nearest even integer
```

### Infinity and NaN

Floats can represent positive infinity, negative infinity, and NaN (“not a number”). NaN is unusual because it is not equal to itself; use `math.isnan()` to test it.

```python
# Special floating-point values

import math

infinity = float("inf")
not_a_number = float("nan")

print(math.isinf(infinity))
print(not_a_number == not_a_number)
print(math.isnan(not_a_number))
```

### Exact decimal arithmetic with `Decimal`

`Decimal` stores decimal values using decimal rules instead of binary floating-point rules. Create values from strings when the original decimal value must be preserved exactly.

```python
# Decimal arithmetic

from decimal import Decimal

first = Decimal("0.1")
second = Decimal("0.2")

print(first + second)
print(Decimal("0.1") * 3)
```

### Decimal precision and context

Decimal operations use a context that controls precision and rounding. `localcontext()` lets one block temporarily use different settings without changing the global context.

```python
# Temporary Decimal precision

from decimal import Decimal, localcontext

with localcontext() as context:
    context.prec = 5
    print(Decimal(1) / Decimal(7))
```

### Exact rational arithmetic with `Fraction`

`Fraction` stores a numerator and denominator and automatically reduces the result. It is useful when a value such as one-third should remain exact.

```python
# Fraction arithmetic

from fractions import Fraction

one_third = Fraction(1, 3)

print(one_third)
print(one_third + Fraction(1, 6))
print(Fraction("0.75"))
print(Fraction(0.333).limit_denominator(10))
```

### Complex numbers: `complex`

A complex number contains a real part and an imaginary part. Python uses `j` for the imaginary unit. Complex numbers support arithmetic but do not support ordering comparisons such as `<` and `>`.

```python
# Complex-number properties

number = 2 + 3j

print(number)
print(number.real)
print(number.imag)
print(abs(number))
print(number.conjugate())
```

```python
# Complex-number arithmetic

number = 2 + 3j

print(number * 2)
print(number ** 2)
print(number / 2)
```

### Numeric comparisons

Comparisons return booleans. Python also supports chained comparisons: `x < y < z` means `x < y and y < z`, while evaluating the middle expression only once.

```python
# Numeric comparisons

x = 2
y = 3
z = 4

print(x < y)
print(x == y)
print(x != y)
print(x < y < z)
print(1 == 2 < 3) # Same meaning as: 1 == 2 and 2 < 3
```

### Booleans behave like integers

`bool` is a subclass of `int`. `True` and `False` are singleton boolean objects that compare equal to `1` and `0`, and they can participate in arithmetic. They are not wrappers placed around integer objects.

```python
# Boolean and integer behavior

print(issubclass(bool, int))
print(True == 1)
print(False == 0)
print(True + True)
print(True * 5)
print(False * 5)
```

### The `math` module

`math` is part of Python’s standard library. `floor()` rounds down, `ceil()` rounds up, and `trunc()` removes the fractional part toward zero.

```python
# Floor, ceiling, and truncation

import math

print(math.floor(4.9))
print(math.ceil(4.1))
print(math.floor(-3.5))
print(math.ceil(-3.5))
print(math.trunc(-2.8))
```

```python
# Common math functions

import math

print(math.sqrt(16))
print(math.factorial(4))
print(math.gcd(8, 12))
print(math.log(1))
print(math.pi)
print(math.pow(2, 3)) # Always returns a float
```

### Generating pseudo-random values

The standard-library `random` module produces pseudo-random values. A seed makes a sequence repeatable, which is useful for testing. `random` is not suitable for passwords or security tokens; use `secrets` for those tasks.

```python
# Pseudo-random values

import random

random.seed(10)
values = [1, 2, 3, 4, 5]

print(random.random())
print(random.randint(1, 10))
print(random.choice(values))
random.shuffle(values)
print(values)
```

## Related Collection: Sets

A `set` is **not a numeric type**. It is a mutable collection of unique hashable objects. It is included here because its union, intersection, and difference operators resemble mathematical set operations.

```python
# Set basics

numbers = {1, 2, 3, 3, 2, 1}

print(numbers) # Duplicates are removed
numbers.add(4)
print(numbers)
```

```python
# Set operations

first = {1, 2, 3}
second = {2, 3, 4}

print(first & second) # Intersection
print(first | second) # Union
print(first - second) # Difference
print(first ^ second) # Symmetric difference
```

```python
# Empty set versus empty dictionary

empty_set = set()
empty_dictionary = {}

print(type(empty_set))
print(type(empty_dictionary))
```

### Quick reference

| Type | Best suited for | Exact? |
|---|---|---|
| `int` | Whole-number arithmetic | Yes |
| `float` | Fast scientific/general decimal calculations | Approximate |
| `Decimal` | Controlled decimal precision, such as financial rules | Yes within its context |
| `Fraction` | Exact ratios | Yes |
| `complex` | Real and imaginary calculations | Components normally use floats |
| `bool` | Logical state and conditions | `True` / `False` |
